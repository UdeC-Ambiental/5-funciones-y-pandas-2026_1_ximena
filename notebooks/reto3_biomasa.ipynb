# Reto 3 – Clasificación de biomasa

**Objetivo:** Clasificar los ecosistemas según su biomasa y variables ambientales asociadas.

Carga el conjunto de datos y explora las variables relevantes.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datos = pd.read_csv("../data/resultados_experimentos.csv")
```

```python
datos.columns
```

```python
def clasificar_biomasa(b):
    
    if b > 250:
        return "alta"
    
    elif 150 <= b <= 250:
        return "media"
    
    else:
        return "baja"
```

```python
datos["nivel_biomasa"] = datos["biomasa"].apply(clasificar_biomasa)
print(datos)
```

```python
datos.head()
```

```python
promedios = datos.groupby("nivel_biomasa")[["temperatura", "humedad", "luz"]].mean()
```

```python
print(promedios)
```

```python
promedios.plot(kind="bar")
plt.title("Variables ambientales por nivel de biomasa")
plt.ylabel("Promedio")
plt.xticks(rotation=0)
plt.show()
```

# Conclusión

Los ecosistemas con biomasa alta presentan mayores niveles de luz y humedad, lo que podría indicar condiciones más favorables para la productividad vegetal.
