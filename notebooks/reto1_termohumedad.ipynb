## Desarrollo del reto
Aplica funciones y análisis según el objetivo. Usa `def` para definir tus funciones.

```python
# Crea aquí tus funciones y análisis de cada reto (Resumen termico)
def resumen_termico(df):

    resumen = pd.DataFrame({
        "Promedio Temperatura": [df["temperatura"].mean()],
        "Máxima Temperatura": [df["temperatura"].max()],
        "Mínima Temperatura": [df["temperatura"].min()],
        "Promedio Humedad": [df["humedad"].mean()],
        "Máxima Humedad": [df["humedad"].max()],
        "Mínima Humedad": [df["humedad"].min()]
    })

    return resumen
resumen_termico(df)
```

## Crea aquí tus funciones y análisis de cada reto (clima, temperatura y humedad)
La función sirve para clasificar el tipo de clima según:

la temperatura (temp)
y la humedad (hum).


```python
def clasificar_clima(temp, hum):

    if temp >= 20 and hum >= 60:
        return "húmedo templado"

    elif temp >= 20 and hum < 60:
        return "seco templado"

    elif temp < 20 and hum >= 60:
        return "húmedo frío"

    else:
        return "seco frío"
    df["clima"] = df.apply(
    lambda fila: clasificar_clima(fila["temperatura"], fila["humedad"]),
    axis=1
)

df.head()
    
```

## Generar tabla de frecuencias
apply() recorre cada fila del DataFrame y ejecuta la función usando:

la temperatura de esa fila
y la humedad de esa fila.

```python
df["clima"] = df.apply(
    lambda fila: clasificar_clima(fila["temperatura"], fila["humedad"]),
    axis=1
)

df.head()
```

## Tabla de frecuencia

```python
tabla_frecuencias = df["clima"].value_counts()

print(tabla_frecuencias)
```
